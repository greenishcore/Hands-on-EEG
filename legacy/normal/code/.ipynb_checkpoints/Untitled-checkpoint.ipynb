```python
import pandas as pd
import numpy as np
import pywt
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("C:\\Users\\a1882\\Desktop\\新建文件夹\\walkfocus_zyy_EPOCFLEX_2023.03.26T15.13.31+08.00.md.bp.csv", header=1)
print(df)
```

```python
ch_info = df.iloc[:, [0] + list(range(3, 35))]
ch_info
```

```python
data_c = ch_info.iloc[:,1:]
data_c
```

```python
df.index
```

```python
df = data_c.T 
# 创建32个子图
fig, axs = plt.subplots(32, figsize=(40, 40))
hex_green = '#00251C'
#hex_orange = '#f56842'
# 绘制每个子图
for i in range(32):
    axs[i].plot(df.iloc[i,:],hex_green)
    axs[i].set_title(f'Channel {df.index[i]}',fontsize=25)


# 添加标题和轴标签
fig.suptitle('32-Channel EEG Data',fontsize=45)
plt.xlabel('Data points',fontsize=35)
plt.ylabel('Amplitude (uV)',fontsize=25)

# 调整子图之间的间距和边距
plt.subplots_adjust(hspace=3)
plt.savefig("channel_eeg.png", dpi=500, bbox_inches='tight')
plt.show()
```

```python
df = data_c.T 
# 创建32个子图
fig, axs = plt.subplots(32, figsize=(90, 90))
hex_green = '#00251C'
# 绘制每个子图
for i in range(32):
    axs[i].plot(df.iloc[i,:],hex_green)
    axs[i].set_title(f'Channel {df.index[i]}')


# 添加标题和轴标签
fig.suptitle('32-Channel EEG Data')
plt.xlabel('Data points')
plt.ylabel('Amplitude (uV)')
plt.rc('xtick', labelsize=25) 
plt.rc('ytick', labelsize=25) 
plt.rc('font', size=30)

# 调整子图之间的间距和边距
plt.subplots_adjust(hspace=1.5)
plt.savefig("channel_eeg.png", dpi=500, bbox_inches='tight')
plt.show()
```
